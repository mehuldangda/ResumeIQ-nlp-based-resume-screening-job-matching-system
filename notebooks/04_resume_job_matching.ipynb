# ResumeIQ — Notebook 04: Resume Screening & Job Matching

In this notebook, we demonstrate the complete end-to-end evaluation pipeline:
1. **Semantic Similarity Scoring (BERT)**
2. **Technical Skill Match Scoring**
3. **Hybrid Compatibility Formula (60% Semantic + 40% Skill)**
4. **Automated Job Recommendations from the Catalog**

## 1. Import Modules and Setup

```python
from pathlib import Path
import sys
import json

# Add project root to sys.path
project_root = Path("..").resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.resume_scanner import compare
from src.job_recommender import recommend_jobs
print("Matching modules imported successfully!")
```

## 2. Define Sample Candidate Profiles & Job Description

```python
candidate_ml = """
Machine Learning Engineer with 3 years experience in Python, PyTorch, Scikit-learn, SQL, and Docker.
Built NLP classification models, deployed REST APIs, and automated data pipelines with Pandas.
"""

candidate_frontend = """
Frontend Web Developer skilled in JavaScript, TypeScript, React, HTML, CSS, and Node.js.
Experienced in responsive UI design and integrating RESTful APIs.
"""

target_jd = """
We are hiring a Machine Learning Engineer.
Requirements: Strong background in Python, Machine Learning, NLP, SQL, Pandas, NumPy, Scikit-learn, and Git.
Experience with Docker and REST APIs is a plus.
"""

print("Candidates and Target JD defined.")
```

## 3. Run Hybrid Resume Screening (`compare`)

```python
candidates = [candidate_ml, candidate_frontend]
results = compare(candidates, target_jd, flag="HuggingFace-BERT")

names = ["Candidate 1 (ML Background)", "Candidate 2 (Frontend Background)"]
for name, res in zip(names, results):
    print("=" * 60)
    print(f"📄 {name}")
    print(f"🎯 Final Match Score:    {res['final_score']}%")
    print(f"🧠 Semantic Similarity:  {res['semantic_score']}%")
    print(f"🛠️ Skill Match Score:    {res['skill_score']}%")
    print(f"✅ Matched Skills:       {', '.join(res['matched_skills'])}")
    print(f"❌ Missing Skills:       {', '.join(res['missing_skills'])}")
```

## 4. Run Job Recommendations Against Database (`recommend_jobs`)

```python
print("Finding job recommendations for Candidate 1 (ML Background):\n")
recommendations = recommend_jobs(candidate_ml)

for rank, job in enumerate(recommendations, 1):
    print(f"Rank #{rank}: {job['job_title']} at {job['company']}")
    print(f"  Match Score:     {job['match_score']}%")
    print(f"  Semantic Score:  {job['semantic_score']}%")
    print(f"  Skill Score:     {job['skill_score']}%")
    print(f"  Skills Matched:  {job['matched_skills']}")
    print(f"  Skills to Learn: {job['missing_skills']}")
    print("-" * 50)
```

## Key Takeaways
- The hybrid scoring algorithm gives high precision because it requires both contextual understanding and hard skill overlap.
- Candidates receive concrete diagnostic feedback on which specific skills they should learn to improve their hiring chances.
