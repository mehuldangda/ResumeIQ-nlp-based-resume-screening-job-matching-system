# ResumeIQ — Notebook 01: Data Exploration

Welcome to the **Data Exploration** notebook for **ResumeIQ**.
In this notebook, we examine the dataset structure, explore the available job listings, inspect the required skills across roles, and understand how resume data is structured.

## 1. Import Libraries and Load Jobs Dataset

```python
import pandas as pd
from pathlib import Path
import sys

# Ensure project root is on Python path
project_root = Path("..").resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.utils import get_data_path
from src.skill_extractor import extract_skills, TECH_SKILLS

# Load jobs catalog
data_path = get_data_path("jobs.csv")
df = pd.read_csv(data_path)
print(f"Loaded {len(df)} job records from: {data_path}")
df.head()
```

## 2. Dataset Structure and Summary

```python
print("DataFrame Info:")
df.info()
print("\nMissing values per column:")
print(df.isnull().sum())
```

## 3. Skill Extraction Across Job Listings
We can apply ResumeIQ's skill extractor to see what technical skills are required in each job description.

```python
df["extracted_skills"] = df["job_description"].apply(extract_skills)
df["skill_count"] = df["extracted_skills"].apply(len)

for idx, row in df.iterrows():
    print(f"📌 Role: {row['job_title']} ({row['company']})")
    print(f"   Skills ({row['skill_count']}): {', '.join(row['extracted_skills'])}")
    print("-" * 60)
```

## 4. Most In-Demand Skills Across Catalog

```python
all_skills = [skill for skills in df["extracted_skills"] for skill in skills]
skill_counts = pd.Series(all_skills).value_counts()

print("Top In-Demand Skills:")
print(skill_counts)
print(f"\nTotal Unique Skills in Catalog: {len(skill_counts)}")
print(f"Total Predefined Skills in Dictionary: {len(TECH_SKILLS)}")
```

## Key Takeaways
- The dataset contains core technical roles (Machine Learning Engineer, Data Analyst, Python Developer, React Developer, Software Developer).
